### 1. 숫자 연산

```python
# 더하기(+)
titanic_train['fare_plus10'] = titanic_train['Fare'] + 10
titanic_train.loc[:, ['Fare', 'fare_plus10']]
```

```python
# 빼기(-)
titanic_train['fare_minus5'] = titanic_train['Fare'] - 5
titanic_train.loc[:, ['Fare', 'fare_minus5']]
```

```python
# 곱하기 (*)
titanic_train['double_fare'] = titanic_train['Fare'] * 2
titanic_train.loc[:, ['Fare', 'double_fare']]
```

```python
# 나누기
titanic_train['fare_divid'] = titanic_train['Fare'] / 2
titanic_train.loc[:, ['Fare', 'fare_divid']]
```

```python
# 나누기(정수만)
titanic_train['fare_divid2'] = titanic_train['Fare'] // 2
titanic_train = titanic_train.astype({'fare_divid2':'int64'})
titanic_train.loc[:, ['Fare', 'fare_divid2']]
```

```python
# 나머지
titanic_train['fare_divid3'] = titanic_train['Fare'] % 2
titanic_train.loc[:, ['Fare', 'fare_divid3']]
```

### 조건문

```python
titanic_train = pd.read_csv('/content/drive/MyDrive/데이터리안/Data/train.csv')
```

```python
# 조건 1개
titanic_train['age_category'] = np.where(titanic_train['Age'] >= 60, 'old', 'young' )
titanic_train.loc[30:35,['Age', 'age_category']]
```

```python
# 조건 2개 이상
filtered = titanic_train[titanic_train['Age'].notna()]

conditions = [
    (filtered['Age'] < 19),
    (filtered['Age'] < 40),
    (filtered['Age'] < 65),
    (filtered['Age'] >= 65),
]

values = ['미성년', '청년', '중장년', '노년']

filtered['age_group'] = np.select(conditions, values)

filtered.loc[24:35,['Age', 'age_group']]
```

```python
filtered.groupby(['Sex', 'age_group'])['PassengerId'].count().reset_index().sort_values(by = ['age_group', 'Sex'])
```
