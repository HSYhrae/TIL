### 1. 데이터 추출

```python
import pandas as pd
```

```python
titanic_train = pd.read_csv('/content/drive/MyDrive/데이터리안/Data/train.csv')
```

```python
# 전체 데이터 추출
titanic_train
```

```python
# 일부 데이터 추출(PassengerId, Name)
# 1. loc
titanic_train.loc[:, ['PassengerId', 'Name']]
```

```python
# 2. iloc
titanic_train.iloc[:, [0, 3]]
```

```python
# 중복 없이 데이터 추출
titanic_train[['Pclass']].drop_duplicates(subset='Pclass')
```

```python
titanic_train[['Survived', 'Pclass']].drop_duplicates(subset=['Survived', 'Pclass'])
```

```python
# 컬럼명 변경하여 추출
titanic_train[['Name', 'Age']].rename(columns={'Name':'이름', 'Age':'나이'})
```

### 2. 데이터 필터링

```python
# 조건에 맞는 데이터 추출
titanic_train[titanic_train['Sex'] == 'male']
```

```python
# 비교 연산자
# 같음
titanic_train[titanic_train['Age'] == 40]
```

```python
# 초과
titanic_train[titanic_train['Age'] > 40]
```

```python
# 이상
titanic_train[titanic_train['Age'] >= 40]
```

```python
# 미만
titanic_train[titanic_train['Age'] < 40]
```

```python
# 이하
titanic_train[titanic_train['Age'] <= 40]
```

```python
# 다름
titanic_train[titanic_train['Age'] != 40]
```

```python
# 논리 연산자
# AND
titanic_train[(titanic_train['Sex'] == 'male') & (titanic_train['Age'] >= 40)]
```

```python
# OR
titanic_train[(titanic_train['Sex'] == 'male') | (titanic_train['Age'] >= 40)]
```

```python
# 여러 데이터 한 번에 추출
# isin()
titanic_train[titanic_train['Pclass'].isin([1, 2])]
```

```python
# between()
titanic_train[titanic_train['PassengerId'].between(100, 200)]
```

```python
titanic_train[(titanic_train['PassengerId'] > 100) & (titanic_train['PassengerId'] < 200)]
```

```python
# 값이 없는/있는 데이터 추출
# isnull(), isna()
titanic_train[titanic_train['Cabin'].isna()]
```

```python
# notnull(), notna()
titanic_train[titanic_train['Cabin'].notna()]
```

```python
# 패턴에 맞는/다른 문자 추출
# 맞는
titanic_train[titanic_train['Name'].str.contains('Mr\.')]
```

```python
# 다른
titanic_train[titanic_train['Name'].str.contains('Mr\.') == False]
```

### 3. 데이터 순서 정렬

```python
# 컬럼 1개 정렬
# 내림차순
titanic_train.sort_values(by='Age', ascending = False)
```

```python
# 오름차순
titanic_train.sort_values(by='Age')
```

```python
# 컬럼 2개 이상 정렬
# 나이는 내림차순, 승객번호는 오름차순
titanic_train.sort_values(by=['Age', 'PassengerId'], ascending = [False, True])
```

```python
# 모두 내림차순
titanic_train.sort_values(by=['Age', 'PassengerId'], ascending = False)
```
